```python
class Node:
    def __init__(self,val):
        self.data = val
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def append(self,newwnode):
        if (self.head== None):
            self.head = newwnode
        else:
            temp = self.head
            while(temp.next):
                temp = temp.next
            temp.next = newwnode

    def insert(self, newwnode, pos):
        if pos < 1:
            raise ValueError("Position must be at least 1")
        if pos == 1:
            newwnode.next = self.head
            self.head = newwnode
            return
        temp = self.head
        for _ in range(pos - 2):
            if temp is None:
                raise IndexError("Position is beyond the end of the list")
            temp = temp.next
        if temp is None:
            raise IndexError("Position is beyond the end of the list")
        newwnode.next = temp.next
        temp.next = newwnode

    def del_neg(self):
        while self.head is not None and self.head.data < 0:
            self.head = self.head.next
        prev = self.head
        current = self.head.next if self.head is not None else None
        while current is not None:
            if current.data < 0:
                prev.next = current.next
            else:
                prev = current
            current = current.next

    def print(self):
        temp = self.head
        print("Linked list")
        while(temp):
            print(temp.data, end  = " ")
            temp = temp.next

list = LinkedList()
list.append(Node(17))
list.append(Node(-12))
list.append(Node(99))
list.print()
```

```python
# Insert at one-based positions; position len(list) + 1 appends.
list.insert(Node(25), 4)
list.insert(Node(-33), 2)
print("\nAfter insert")
list.print()
```

```python
# Remove all negative-valued nodes.
print("\nDelete negative")
list.del_neg()
list.print()
```
