```python
# Q. 1 Create a Stack with following operations
# 1. Push
# 2. Pop
# 3. Display

class Node:
    def __init__(self, data, next: "Node | None" = None):
        self.data = data
        self.next = next

class Stack:
    def __init__(self, top: Node | None = None):
        self.top = top

    def isEmpty(self) -> bool:
        return self.top is None

    def push(self, new_node):
        temp = Node(new_node)
        temp.next = self.top
        self.top = temp

    def pop(self):
        if self.top is None:
            print('Stack is Empty')
            return None
        temp = self.top
        print('Popped Element :- ', temp.data)
        self.top = temp.next
        return temp.data

    def display(self):
        if self.top is None:
            print('Stack is Empty')
        else:
            print('Stack Contains...')
            temp = self.top
            while temp is not None:
                print(temp.data, end=' -> ')
                temp = temp.next
            print('Null')

    def reverse_string(self, data: str) -> str:
        stack = []
        for ch in data:
            stack.append(ch)
        result = ""
        while stack:
            result += stack.pop()
        return result

    def reverse_stack(self, s1: "Stack", s2: "Stack"):
        while not s1.isEmpty():
            s2.push(s1.pop())

s = Stack()
s.push(1)
s.push(2)
s.push(3)
s.push(4)
s.push(5)
s.display()
print('After a Pop Operation :- ')
s.pop()
s.display()
```

```python
print(s.reverse_string('Yogesh'))
```

```python
s1 = Stack()
s1.push(10)
s1.push(20)
s1.push(30)
s2 = Stack()
s1.reverse_stack(s1, s2)
print('Reversed stack contents:')
s2.display()
```

```python
class Node:
    def __init__(self, data, next: "Node | None" = None):
        self.data = data
        self.next = next

class Queue:
    def __init__(self):
        self.front: Node | None = None
        self.rear: Node | None = None

    def enqueue(self, data):
        temp = Node(data)
        if self.rear is None:
            self.front = self.rear = temp
        else:
            self.rear.next = temp
            self.rear = temp

    def dequeue(self):
        if self.front is None:
            print("Queue is Empty")
            return None
        temp = self.front
        print("Dequeued Element:", temp.data)
        self.front = temp.next
        if self.front is None:
            self.rear = None
        return temp.data

    def display(self):
        if self.front is None:
            print("Queue is Empty")
        else:
            print("Queue Contains...")
            temp = self.front
            while temp is not None:
                print(temp.data, end=" -> ")
                temp = temp.next
            print("None")

    def reverse_queue(self):
        stack = []
        temp = self.front
        while temp is not None:
            stack.append(temp.data)
            temp = temp.next
        self.front = self.rear = None
        while stack:
            self.enqueue(stack.pop())

    def merge_queue(self, q2):
        temp = q2.front
        while temp is not None:
            self.enqueue(temp.data)
            temp = temp.next

    def identical(self, q2):
        temp1 = self.front
        temp2 = q2.front
        while temp1 is not None and temp2 is not None:
            if temp1.data != temp2.data:
                return False
            temp1 = temp1.next
            temp2 = temp2.next
        return temp1 is None and temp2 is None
```

```python
q1 = Queue()
q1.enqueue(10)
q1.enqueue(20)
q1.enqueue(30)
q1.enqueue(40)
q1.display()
```

```python
q1.reverse_queue()
q1.display()
```

```python
q2 = Queue()
q2.enqueue(10)
q2.enqueue(20)
q2.enqueue(30)
q3 = Queue()
q3.enqueue(10)
q3.enqueue(20)
q3.enqueue(30)
print(q1.identical(q2))
print(q2.identical(q3))
```
